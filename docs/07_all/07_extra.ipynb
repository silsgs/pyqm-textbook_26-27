# Problem 1

Problem on time evolution of a given many-body hamiltonian:

Consider a one-dimensional chain of spin-1/2 particles with nearest-neighbor interactions described by the Heisenberg Hamiltonian:
\[ H = J \sum_{i=1}^{N-1} \left( S_i^x S_{i+1}^x + S_i^y S_{i+1}^y + S_i^z S_{i+1}^z \right) \]
where \( S_i^{x,y,z} \) are the spin operators for the \( i \)-th particle, \( J \) is the coupling constant, and \( N \) is the number of particles in the chain.

Your task is to simulate the time evolution of an initial state \( |\psi(0)\rangle \) under this Hamiltonian using the time-dependent Schrödinger equation:
\[ i\hbar \frac{d}{dt} |\psi(t)\rangle = H |\psi(t)\rangle \]

# Solution 1
```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.linalg import expm

# Define parameters
N = 4  # Number of spins
J = 1.0  # Coupling constant
hbar = 1.0  # Reduced Planck's constant
dt = 0.1  # Time step
t_max = 10.0  # Maximum time
num_steps = int(t_max / dt)
# Define spin operators
Sx = np.array([[0, 1], [1, 0]]) / 2
Sy = np.array([[0, -1j], [1j, 0]]) / 2
Sz = np.array([[1, 0], [0, -1]]) / 2
I = np.eye(2)
# Construct the Heisenberg Hamiltonian
H = np.zeros((2**N, 2**N), dtype=complex)
for i in range(N - 1):
    Si_x = np.kron(np.kron(np.eye(2**i), Sx), np.eye(2**(N - i - 1)))
    Si_y = np.kron(np.kron(np.eye(2**i), Sy), np.eye(2**(N - i - 1)))
    Si_z = np.kron(np.kron(np.eye(2**i), Sz), np.eye(2**(N - i - 1)))
    Si1_x = np.kron(np.kron(np.eye(2**(i + 1)), Sx), np.eye(2**(N - i - 2)))
    Si1_y = np.kron(np.kron(np.eye(2**(i + 1)), Sy), np.eye(2**(N - i - 2)))
    Si1_z = np.kron(np.kron(np.eye(2**(i + 1)), Sz), np.eye(2**(N - i - 2)))
    H += J * (np.dot(Si_x, Si1_x) + np.dot(Si_y, Si1_y) + np.dot(Si_z, Si1_z))
# Initial state: all spins up
psi_0 = np.zeros((2**N, 1), dtype=complex)
psi_0[0] = 1.0
# Time evolution
times = np.linspace(0, t_max, num_steps)
magnetizations = []
for t in times:
    U = expm(-1j * H * t / hbar)  # Time evolution operator
    psi_t = np.dot(U, psi_0)  # Evolve the state
    # Calculate magnetization
    Mz = 0
    for i in range(N):
        Si_z = np.kron(np.kron(np.eye(2**i), Sz), np.eye(2**(N - i - 1)))
        Mz += np.vdot(psi_t, np.dot(Si_z, psi_t)).real
    magnetizations.append(Mz / N)
# Plot results
plt.plot(times, magnetizations)
plt.xlabel('Time')
plt.ylabel('Average Magnetization <Mz>')
plt.title('Time Evolution of Magnetization in Heisenberg Chain')
plt.grid()
plt.show()
```

## 4.1 Coherent states

A coherent state of a one-dimensional simple harmonic oscillator is defined to be an eigenstate of the non-Hermitian (i.e. $a \neq a^{\dagger}$) annihilation operator $a$. 

$$ a|\alpha\rangle = \lambda |\alpha\rangle $$

where $\lambda$, because of the non-Hermiticity of $a$, is a complex number. Coherent states are specific linear combinations of harmonic oscillator eigenfunction that produce Gaussian wave packets that do not spread in time. Moreover, if the uncertainties in position and time are equal, then the resulting wave packet would be as close a representation of a classical particle as could be obtained whithin the bounds of the uncertainty principle.

1. Show that a coherent state can also be obtained by appliying the translation (finite displacement) operator:

$$ e^{\alpha a^\dagger - \alpha^{*} a} | 0 \rangle = | \alpha \rangle $$

Solution:
```python
import numpy as np
from scipy.linalg import expm

# Define the annihilation and creation operators
a = np.array([[0, 1], [0, 0]])
adag = np.array([[0, 0], [1, 0]])

# Define the ground state |0>
psi_0 = np.array([[1], [0]])

# Define the translation operator
alpha = 1.0 + 1.0j
T = expm(alpha * adag - np.conj(alpha) * a)

# Apply the translation operator to the ground state
psi_alpha = np.dot(T, psi_0)
print("Coherent state |alpha>:\n", psi_alpha)
```
