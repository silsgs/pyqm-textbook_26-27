# Part 2. Introduction to PyTorch

PyTorch is an open-source library for machine learning and deep learning. It provides a flexible and efficient platform for building and training neural networks, and it is widely used in both research and industry.

There are many Python libraries available for developing machine learning algorithms, e.g. TensorFlow, Keras, Scikit-learn, and PyTorch. Among the most commonly used in both research and industry are TensorFlow and PyTorch.

The main advantage of PyTorch is its dynamic computation graph, which allows for more flexibility in building and modifying neural networks compared to static computation graphs used in other frameworks like TensorFlow.

**Static vs. dynamic computation graphs**

- **Static Computation Graphs**: Used in frameworks like TensorFlow (prior to eager execution). The computation graph is defined before running the model, and it cannot be changed during execution.
- **Dynamic Computation Graphs**: Used in PyTorch. The computation graph is built on-the-fly as operations are performed, allowing for more flexibility in modifying the network architecture during training. The main advantage is that it enables more adaptive and flexible model design, especially for complex tasks where the network structure may need to change dynamically, as is in the case of recurrent neural networks (RNNs) or models with conditional computation (conditional computation means that certain parts of the network are activated only under specific conditions).


**Key Features of PyTorch**

Within the main features that PyTorch offers, the most relevant ones include:

- **Dynamic Computation Graphs**: PyTorch uses dynamic computation graphs, which allow you to modify the graph on the fly. 
- **Tensor Computation**: PyTorch provides a powerful N-dimensional array (tensor) library that supports GPU acceleration, making it efficient for large-scale numerical computations.
- **Autograd**: PyTorch includes an automatic differentiation library called Autograd, which simplifies the process of computing gradients for backpropagation.


**Basic concepts in Machine Learning**

Machine Learning aims at developing algorithms that allow computers to learn patterns and make predictions or decisions based on data. Many of nowaday's applications, such as image recognition, natural language processing, and recommendation systems, rely heavily on machine learning techniques. It is based basically on interpolation and generalization from observed data to unseen data. Because of this, having a large enough and representative dataset is crucial for the success of machine learning models. 

*Interpolation* refers to the process of estimating unknown values that fall within the range of known data points. In machine learning, interpolation is used to make predictions for new data points based on the patterns learned from the training data.

*Generalization*, on the other hand, refers to the ability of a machine learning model to perform well on unseen data, not just the training data. A model that generalizes well can make accurate predictions on new, previously unseen examples, which is the ultimate goal of most machine learning tasks.

A typical workflow in machine learning involves the following steps:
1. **Data Collection**: Gather and prepare a dataset that is representative of the problem you want to solve. (e.g., calculating tensorial properties of materials, scalars asssociated with them, coefficients, etc.)
2. **Data Preprocessing**: Clean and preprocess the data, including handling missing values, normalizing features, and splitting the data into training and testing sets. (e.g., normalizing tensor values, handling missing coefficients, etc.)
3. **Model Selection**: Choose an appropriate machine learning model based on the problem type (e.g., regression, classification).
4. **Training**: Train the model on the training dataset by optimizing its parameters to minimize the error. (e.g., using gradient descent to update model parameters.)
5. **Evaluation**: Evaluate the model's performance on the testing dataset to assess its generalization ability.
6. **Hyperparameter Tuning**: Adjust hyperparameters to improve model performance. (e.g., learning rate, batch size, number of layers, etc.)
7. **Deployment**: Deploy the trained model to make predictions on new, unseen data. (e.g., using the model to predict properties of new materials.)

Very important steps are the data related, collection and normalization procedures. And testing of hyperparameters, which is crucial to ensure the model performs well on unseen data. Then, blind testing is essential for validating the accuracy and robustness of the final model.


## 2.2 Objects in PyTorch
In PyTorch, the primary object is the **tensor**, which is a multi-dimensional array similar to NumPy arrays but with additional capabilities for GPU acceleration and automatic differentiation. Tensors are the building blocks for creating and manipulating data in PyTorch.

PyTorch provides a wide range of operations for creating, manipulating, and performing computations on tensors. These operations include basic arithmetic, linear algebra, and more advanced functions for deep learning.

Tensors can be created from Python lists or NumPy arrays, and have various methods for initialization, such as `torch.ones`, `torch.zeros`, and `torch.rand`.

Here are some common ways to create tensors in PyTorch:

- From Python lists: `torch.tensor([1, 2, 3])`
- From NumPy arrays: `torch.from_numpy(numpy_array)`
- Using built-in initialization methods: `torch.ones`, `torch.zeros`, `torch.rand`, `torch.randn`

Examples:



````python
import torch
# Creating tensors from Python lists
tensor_from_list = torch.tensor([1, 2, 3])
print("Tensor from Python list:")
print(tensor_from_list)

# Creating tensors from NumPy arrays
import numpy as np
numpy_array = np.array([1, 2, 3])
tensor_from_numpy = torch.from_numpy(numpy_array)
print("Tensor from NumPy array:")
print(tensor_from_numpy)

# Using built-in initialization methods
ones_tensor = torch.ones(3)
zeros_tensor = torch.zeros(3)
rand_tensor = torch.rand(3)
randn_tensor = torch.randn(3)

print("Tensor of ones:")
print(ones_tensor)
print("Tensor of zeros:")
print(zeros_tensor)
print("Random tensor:")
print(rand_tensor)
print("Random normal tensor:")
print(randn_tensor)
````

We can always convert any NumPy array to a PyTorch tensor using `torch.from_numpy()`. This allows seamless integration between NumPy and PyTorch, enabling you to leverage the strengths of both libraries in your computations.

In the same way, you can convert a PyTorch tensor back to a NumPy array using the `.numpy()` method. This bidirectional conversion ensures that you can easily switch between NumPy and PyTorch depending on your computational needs.

Example:


````python
import torch
import numpy as np

# Converting a NumPy array to a PyTorch tensor
numpy_array = np.array([1, 2, 3])
tensor_from_numpy = torch.from_numpy(numpy_array)
print("Tensor from NumPy array:")
print(tensor_from_numpy)

# Converting a PyTorch tensor back to a NumPy array
tensor = torch.tensor([4, 5, 6])
numpy_from_tensor = tensor.numpy()
print("NumPy array from PyTorch tensor:")
print(numpy_from_tensor)
````

## 2.3 Autograd

Autograd is PyTorch's automatic differentiation engine that powers neural network training. It records operations performed on tensors to compute gradients automatically.

### 2.3.1 Enabling Autograd

To enable autograd for a tensor, set the `requires_grad` attribute to `True`:

```python
import torch

x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
print(x.requires_grad)  # True
```

### 2.3.2 Computing Gradients

Once autograd is enabled, you can perform operations on the tensor and compute gradients using the `backward()` method:

```python
y = x.sum()
y.backward()
print(x.grad)  # Gradient of y with respect to x
```

## 2.4 A general workflow in PyTorch

A short guide on how to implement a general Machine Learning workflow in PyTorch.

1. Load Data: You can load your dataset into PyTorch, typically using `torch.utils.data.DataLoader`, or you can load it directly into tensors using `torch.tensor`.

In this example we will load pre-existing data from available datasets in PyTorch. 



````python
# Download trainig data from open datasets
from torchvision import datasets, transforms
training_data = datasets.FashionMNIST(
    root="data",
    train=True,  # this flag defines this set as the training
    download=True,
    transform=transforms.ToTensor()
)

test_data = datasets.FashionMNIST(
    root="data",
    train=False,  # this flag defines this set as the test
    download=True,
    transform=transforms.ToTensor()
)
````

2. Create DataLoaders

Once the datasets are loaded, we can create DataLoaders to efficiently iterate over the data in batches. Batches mean that instead of processing the entire dataset at once, we process a smaller subset of the data, which helps with memory management and can speed up training. 

````python
from torch.utils.data import DataLoader
batch_size = 64
train_loader = DataLoader(training_data, batch_size=batch_size, shuffle=True) # here we shuffle the training data to ensure the model sees the data in a different order each epoch
test_loader = DataLoader(test_data, batch_size=batch_size, shuffle=False) # here shuffle is set to False because we typically don't need to shuffle the test data

for X, y in test_loader:
    print(f"Shape of X [N, C, H, W]: {X.shape}")
    print(f"Shape of y: {y.shape} {y.dtype}")
    break
````

3. Creating a Neural Network Model

Now that we have our data prepared and loaded into DataLoaders, the next step is to define a neural network model. In PyTorch, we typically define a model by subclassing `torch.nn.Module` and defining the layers and the forward pass.

Here is an example of a simple feedforward neural network for the FashionMNIST dataset:


````python
import torch
import torch.nn as nn

device = torch.accelerator.current_accelerator().type if torch.accelerator.is_available() else "cpu"
print(f"Using {device} device")

# Defining the model
class FashionMNISTModel(nn.Module):
    def __init__(self):
        super(FashionMNISTModel, self).__init__()
        self.flatten = nn.Flatten()  # Flatten the 28x28 input images into a 1D vector
        self.linear_relu_stack = nn.Sequential(  # A stack of linear layers with ReLU activations
            nn.Linear(28*28, 512),
            nn.ReLU(),
            nn.Linear(512, 512),
            nn.ReLU(),
            nn.Linear(512, 10)
        )

    def forward(self, x):
        x = self.flatten(x)
        logits = self.linear_relu_stack(x)
        return logits

model = FashionMNISTModel().to(device)
print(model)
````

4. Optimizing the Model

Once we have defined our model, the next step is to optimize it. This involves defining a loss function and an optimizer. The loss function measures how well the model is performing, and the optimizer updates the model's parameters to minimize the loss.

Here is an example using the cross-entropy loss and the Adam optimizer:


````python
import torch.optim as optim

# Define the loss function
loss_fn = nn.CrossEntropyLoss()

# Define the optimizer
optimizer = optim.Adam(model.parameters(), lr=1e-3)
````

The loss function is a very important concept in training neural networks. It measures how well the model's predictions match the true labels. During training, the model aims to minimize this loss by adjusting its parameters through the optimization process. In PyTorch, common loss functions include `nn.CrossEntropyLoss` for classification tasks and `nn.MSELoss` for regression tasks.
Usually, the model's performance is evaluated using the same loss function on a separate validation set to monitor its progress and prevent overfitting.

Now we will define the training loop for our model using the loss function and optimizer we have set up. This loop will iterate over the training data, compute the loss, perform backpropagation, and update the model's parameters accordingly.

````python
def train(dataloader, model, loss_fn, optimizer):
    size = len(dataloader.dataset)
    model.train()
    for batch, (X, y) in enumerate(dataloader):
        X, y = X.to(device), y.to(device)

        # Compute prediction error
        pred = model(X)
        loss = loss_fn(pred, y)

        # Backpropagation
        loss.backward()
        optimizer.step()
        optimizer.zero_grad()

        if batch % 100 == 0:
            loss, current = loss.item(), (batch + 1) * len(X)
            print(f"loss: {loss:>7f}  [{current:>5d}/{size:>5d}]")
````

We will need to check the model's performance on a validation set to ensure it is learning correctly and not overfitting to the training data. This involves running the model in evaluation mode and computing the loss and accuracy on the validation data.



````python
def test(dataloader, model, loss_fn):
    size = len(dataloader.dataset)
    num_batches = len(dataloader)
    model.eval()
    test_loss, correct = 0, 0
    with torch.no_grad():
        for X, y in dataloader:
            X, y = X.to(device), y.to(device)
            pred = model(X)
            test_loss += loss_fn(pred, y).item()
            correct += (pred.argmax(1) == y).type(torch.float).sum().item()
    test_loss /= num_batches
    correct /= size
    print(f"Test Error: \n Accuracy: {(100*correct):>0.1f}%, Avg loss: {test_loss:>8f} \n")
````

````python
epochs = 5
for t in range(epochs):
    print(f"Epoch {t+1}\n-------------------------------")
    train(train_loader, model, loss_fn, optimizer)
    test(test_loader, model, loss_fn)
print("Done!")
````

We can see the model's performance on the test set after training for the specified number of epochs. The accuracy and average loss provide an indication of how well the model has learned to classify the FashionMNIST images.
