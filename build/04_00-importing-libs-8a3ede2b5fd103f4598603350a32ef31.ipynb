# Part 0. Importng Python Libraries

As we have seen in previous lectures, Python runs code in modules called libraries. These libraries contain pre-written code that we can use to perform various tasks without having to write everything from scratch.

As we saw last week, importing a library means loading all its functions and classes into our current Python environment. Enabling the use of its objects, functions, and classes in our code. 

There are some specific libraries which have become the standard tools for scientific computing in Python. For instance, Numpy is widely used for numerical computations, Scipy for scientific and technical computing, and Matplotlib for data visualization.

There are many other libraries available for different purposes, but among the most commonly used libraries for different tasks is summarized below:

- **Numpy**: very efficient for *numerical computations*
- **Scipy**: includes specialized modules for *scientific and technical computing*
- **Matplotlib/Seaborn**: *Data visualization*
- **Pandas**: *Data manipulation and analysis*
- **Scikit-learn**: *Machine learning*, it provides powerful frameworks for classification, regression and clustering tasks.
- **TensorFlow/PyTorch**: *Deep learning* and *Artificial Intelligence*
- **PyTorch**: *Deep learning* and *Artificial Intelligence*, builds dynamic computational graphs for flexible model development.

Today we will explore more in detail NumPy, and Matplotlib-based libraries. We will take next session to go deep with Scipy.

We need first to install, and then, import them in our code. **Importing a library means making available all its functions and classes.**

They are open-source and have extensive documentation and community support. They are also compatible with other popular Python libraries such as Pandas, Scikit-learn, and TensorFlow.

## 0.1 Installing Libraries
We might have already installed these libraries if you are using the Anaconda environment we created the first week. But in case you need to install them, we can do so using conda or pip:

```bash
conda install numpy scipy matplotlib pandas
pip install numpy scipy matplotlib pandas
```

This applies to any other library you might want to install in the future. Just replace the library names with the ones you want to install. 
You can also install specific versions of the libraries by specifying the version number. For example, to install version 1.21.0 of Numpy, you can use:

```bash
### This is just an example
# This code runs in the terminal, not in Python
conda install numpy=1.21.0 scipy=1.7.0 matplotlib=3.4.2 pandas=1.3.0
# or
pip install numpy==1.21.0 scipy==1.7.0 matplotlib==3.4.2 pandas==1.3.0
```

We will also install PyTorch, another popular library for machine learning and deep learning. It can be installed using conda or pip as well:

```bash
conda install pytorch torchvision torchaudio -c pytorch
pip install torch torchvision torchaudio
```

## 0.2 Importing Libraries
Once the libraries are installed, since they are not included in the standard Python library, we always need to import them in our Python script or Jupyter Notebook using the following commands:

```python
# This code runs in Python
import numpy as np
import scipy as sp
import matplotlib.pyplot as plt
```

Here, we are importing Numpy as `np`, Scipy as `sp`, and Matplotlib's `pyplot` module as `plt`. This is a common convention in the Python community and makes it easier to use the libraries in our code.

If you don't load them in the beginning of your code, you will get an error when trying to use any of their functions or classes. The error will look like this:

```python
NameError: name 'np' is not defined
```
